```python
def simplex(c, A, b):
    m, n = A.shape

    A = np.hstack([A, np.eye(m)])
    c = np.concatenate([c, np.zeros(m)])

    basis = list(range(n, n + m))
    nonbasis = list(range(n))
    
    while True:
        B = A[:, basis]
        N = A[:, nonbasis]
        cb = c[basis]
        cn = c[nonbasis]

        B_inv = np.linalg.inv(B)
        xb = B_inv @ b
        r = cn - (cb @ B_inv) @ N

        if np.all(r <= 1e-9):
            break

        j = np.argmax(r)
        d = B_inv @ A[:, nonbasis[j]]

        if np.all(d <= 1e-9):
            raise Exception("Unbounded problem")

        l = np.argmin(np.divide(xb, d, out=np.full_like(xb, np.inf), where=d > 1e-9))

        basis[l], nonbasis[j] = nonbasis[j], basis[l]

    B_inv = np.linalg.inv(A[:, basis])
    x = np.zeros(len(c))
    x[basis] = B_inv @ b

    return x[:n], c.dot(x)
```

```python
c = np.array([2, 3, 1])
A = np.array([
    [1, 1, 1],
    [2, 1, 3],
    [1, 2, 1]
])
b = np.array([10, 15, 12])

x, r = simplex(c, A, b)
print(x, r)
```
